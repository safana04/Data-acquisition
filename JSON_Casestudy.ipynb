Task 2


```python
gdp = pd.read_excel('/content/gdp.xlsx')

display(gdp.head())

print("Column Names:", gdp.columns.tolist())
print("\nData Types:")
print(gdp.dtypes)
```

Task 3



```python
import json

with open('/content/internet_users.json') as file:
    internet_data = json.load(file)

internet = pd.DataFrame(internet_data)

display(internet.head())

print(internet.dtypes)
```

Task 4



```python
import xml.etree.ElementTree as ET

tree = ET.parse('/content/literacy_rate.xml')
root = tree.getroot()

literacy_data = []

for country in root.findall('country'):
    name = country.get('name')
    rate = country.find('literacy_rate').text

    literacy_data.append({
        'country': name,
        'literacy_rate': float(rate)
    })

literacy = pd.DataFrame(literacy_data)

display(literacy.head())
```

Task 5



```python
datasets = {
    'Population': population,
    'GDP': gdp,
    'Internet': internet,
    'Literacy': literacy
}

for name, df in datasets.items():

    print("\nDataset:", name)
    print("Rows and Columns:", df.shape)
    print("Column Names:", df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Records:")
    print(df.duplicated().sum())
```

Task 6


```python
gdp.rename(columns={'Country': 'country'}, inplace=True)

literacy.rename(columns={'name': 'country'}, inplace=True)
```

Task 7

```python
population['population'] = pd.to_numeric(population['population'])

gdp['GDP'] = pd.to_numeric(gdp['GDP'])

internet['internet_users'] = pd.to_numeric(internet['internet_users'])

literacy['literacy_rate'] = pd.to_numeric(literacy['literacy_rate'])
```

Task 8

```python
gdp.rename(columns={'Country': 'country'}, inplace=True)
data = pd.merge(population, gdp, on='country', how='inner')

data = pd.merge(data, internet, on='country', how='inner')

data = pd.merge(data, literacy, on='country', how='inner')

display(data.head())

print("Shape:", data.shape)
```

Task 9

```python
data['internet_penetration'] = (
    data['internet_users'] / data['population'] * 100
).round(2)
```

```python
data['gdp_per_capita'] = (
    data['GDP'] * 1000000 / data['population']
).round(2)
```

```python
display(data.head())
```

Task 10

```python
import sqlite3

conn = sqlite3.connect('country_statistics.db')

data.to_sql(
    'country_statistics',
    conn,
    if_exists='replace',
    index=False
)

print("Data loaded successfully!")
```

Task 11

```python
result = pd.read_sql_query(
    "SELECT * FROM country_statistics LIMIT 5",
    conn
)

display(result)
```

Part F – Data Analysis Using SQL


11 Display the Top 5 countries by GDP.


```python
query = """
SELECT country, GDP
FROM country_statistics
ORDER BY GDP DESC
LIMIT 5
"""

pd.read_sql_query(query, conn)
```

12 Find the average literacy rate.


```python
query = """
SELECT AVG(literacy_rate) AS average_literacy_rate
FROM country_statistics
"""

pd.read_sql_query(query, conn)
```

13 Display all countries having literacy rate above 90%.

```python
query = """
SELECT country, literacy_rate
FROM country_statistics
WHERE literacy_rate > 90
"""

pd.read_sql_query(query, conn)
```

14 Find the Top 5 countries with the highest internet penetration.


```python
query = """
SELECT country, internet_penetration
FROM country_statistics
ORDER BY internet_penetration DESC
LIMIT 5
"""

pd.read_sql_query(query, conn)
```

15 Calculate the total population.


```python
query = """
SELECT SUM(population) AS total_population
FROM country_statistics
"""

pd.read_sql_query(query, conn)
```

16 Find the country with the highest GDP per capita.  


```python
query = """
SELECT country, gdp_per_capita
FROM country_statistics
ORDER BY gdp_per_capita DESC
LIMIT 1
"""

pd.read_sql_query(query, conn)
```
